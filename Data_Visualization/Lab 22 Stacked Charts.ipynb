### Task 1: Stacked Chart for Composition of Job Satisfaction Across Age Groups


##### 1. Stacked Chart of Median `JobSatPoints_6` and `JobSatPoints_7` for Different Age Groups


Visualize the composition of job satisfaction scores (`JobSatPoints_6` and `JobSatPoints_7`) across various age groups. This will help in understanding the breakdown of satisfaction levels across different demographics.



```python
##Write your code here

import pandas as pd
import matplotlib.pyplot as plt

# Group by Age and calculate the median for the specific columns
df_grouped = df.groupby('Age')[['JobSatPoints_6', 'JobSatPoints_7']].median()

# Plot stacked bar chart
df_grouped.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Median JobSatPoints_6 and JobSatPoints_7 by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Median Score')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Job Satisfaction Metrics')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of `JobSatPoints_6` and `JobSatPoints_7` for Employment Status


Create a stacked chart to compare job satisfaction (`JobSatPoints_6` and `JobSatPoints_7`) across different employment statuses. This will show how satisfaction varies by employment type.


```python
##Write your code here

# Group by Employment and calculate the median
df_emp = df.groupby('Employment')[['JobSatPoints_6', 'JobSatPoints_7']].median()

# Plot stacked bar chart
df_emp.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Median JobSatPoints_6 and JobSatPoints_7 by Employment Status')
plt.xlabel('Employment Status')
plt.ylabel('Median Score')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Job Satisfaction Metrics')
plt.tight_layout()
plt.show()
```

### Task 2: Stacked Chart for Compensation and Job Satisfaction by Age Group


##### This stacked chart visualizes the composition of compensation (`ConvertedCompYearly`) and job satisfaction (`JobSatPoints_6`) specifically for respondents aged 30-35.


```python
# Run this cell first to check what age categories actually exist in your data
print(df['Age'].unique())
```

```python
##Write your code here

# Replace '30-35 years old' with the actual category string found from df['Age'].unique()
df_filtered = df[df['Age'] == '25-34 years old']

df_filtered.groupby('MainBranch')[['ConvertedCompYearly', 'JobSatPoints_6']].median().plot(
    kind='bar', stacked=True, figsize=(10, 6)
)
plt.title('Compensation and Job Satisfaction for Selected Age Group')
plt.xlabel('Main Branch')
plt.ylabel('Values')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

##### Stacked Chart of Median Compensation and Job Satisfaction Across Age Group


Compare the median compensation and job satisfaction metrics across different age groups. This helps visualize how compensation and satisfaction levels differ by age.


```python
##Write your code here

df_age_comp = df.groupby('Age')[['ConvertedCompYearly', 'JobSatPoints_6']].median()

df_age_comp.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Median Compensation and Job Satisfaction Across Age Groups')
plt.xlabel('Age Group')
plt.ylabel('Values')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

### Task 3: Comparing Data Using Stacked Charts


##### 1. Stacked Chart of Preferred Databases by Age Group




Visualize the top databases that respondents from different age groups wish to learn. Create a stacked chart to show the proportion of each database in each age group.


```python
##Write your code here

# Create dummy columns from the semicolon-separated database column, then concatenate with 'Age'
db_dummies = df['DatabaseWantToWorkWith'].str.get_dummies(sep=';')
db_dummies['Age'] = df['Age']

# Group by 'Age' and sum the dummy variables, then plot
df_db = db_dummies.groupby('Age').sum()

df_db.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title('Preferred Databases by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart of Employment Type by Job Satisfaction


Analyze the distribution of employment types within each job satisfaction level using a stacked chart. This will provide insights into how employment types are distributed across various satisfaction ratings.


```python
##Write your code here


# Create a crosstab between Employment and JobSat, then normalize to show proportions or plot directly
ct = pd.crosstab(df['JobSat'], df['Employment'], normalize='index')

ct.plot(kind='bar', stacked=True, figsize=(10, 6))
plt.title('Distribution of Employment Types within Each Job Satisfaction Level')
plt.xlabel('Job Satisfaction')
plt.ylabel('Proportion')
plt.xticks(rotation=0)
plt.legend(title='Employment Status', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Task 4: Exploring Technology Preferences Using Stacked Charts


##### 1. Stacked Chart for Preferred Programming Languages by Age Group


Analyze how programming language preferences (`LanguageAdmired`) vary across age groups.


```python
##Write your code here

# Create dummy columns from the semicolon-separated LanguageAdmired column, then join with Age
lang_dummies = df['LanguageAdmired'].str.get_dummies(sep=';')
lang_dummies['Age'] = df['Age']

# Group by Age, sum the counts, and plot a stacked bar chart
df_lang = lang_dummies.groupby('Age').sum()

df_lang.plot(kind='bar', stacked=True, figsize=(12, 7))
plt.title('Preferred Programming Languages by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', ncol=2)
plt.tight_layout()
plt.show()
```

##### 2. Stacked Chart for Technology Adoption by Employment Type


Explore how admired platforms (`PlatformAdmired`) differ across employment types (e.g., full-time, freelance)


```python
##Write your code here

# Create dummy columns and group by Employment
platform_dummies = df['PlatformAdmired'].str.get_dummies(sep=';')
platform_dummies['Employment'] = df['Employment']

# Sum and transpose, then filter to show only the top 10 most popular platforms to avoid overcrowding
df_platform = platform_dummies.groupby('Employment').sum().T
df_platform['Total'] = df_platform.sum(axis=1)
df_platform = df_platform.sort_values(by='Total', ascending=False).drop(columns=['Total']).head(10)

# Plot clean horizontal stacked bar chart
df_platform.plot(kind='barh', stacked=True, figsize=(10, 8))
plt.title('Top 10 Admired Platforms by Employment Type')
plt.xlabel('Count')
plt.ylabel('Platform')
plt.gca().invert_yaxis()  # Put the highest count at the top
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
