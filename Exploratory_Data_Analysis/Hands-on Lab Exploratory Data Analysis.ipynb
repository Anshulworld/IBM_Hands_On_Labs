#### Step 2: Load and Preview the Dataset
Load the dataset from the provided URL. Use df.head() to display the first few rows to get an overview of the structure.


```python
# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Set pandas option to display all columns
pd.set_option('display.max_columns', None)

# Display the first few rows of the dataset
df.head()
```

#### Step 3: Handling Missing Data


Identify and manage missing values in critical columns such as `Employment`, `JobSat`, and `RemoteWork`. Implement a strategy to fill or drop these values, depending on the significance of the missing data.


```python
## Write your code here

# Check missing values in critical columns
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())

# Drop rows where critical columns have missing values, as imputation is not reliable here
df = df.dropna(subset=['Employment', 'JobSat', 'RemoteWork'])

# Verify that missing values have been removed
print(df[['Employment', 'JobSat', 'RemoteWork']].isnull().sum())
```

#### Step 4: Analysis of Experience and Job Satisfaction


Analyze the relationship between years of professional coding experience (`YearsCodePro`) and job satisfaction (`JobSat`). Summarize `YearsCodePro` and calculate median satisfaction scores based on experience ranges.

- Create experience ranges for `YearsCodePro` (e.g., `0-5`, `5-10`, `10-20`, `>20` years).

- Calculate the median `JobSat` for each range.

- Visualize the relationship using a bar plot or similar visualization.


```python
## Write your code here

# Convert YearsCodePro to numeric, coercing non-numeric values (like 'More than 50 years' or 'Less than 1 year') to NaN
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# Depending on how JobSat is encoded (categorical vs numeric), map it to a numeric scale if needed
# If JobSat contains text categories, map them to scores. (Adjust dictionary keys based on your unique categories using df['JobSat'].unique())
# jobsat_mapping = {'Very dissatisfied': 1, 'Dissatisfied': 2, 'Neither satisfied nor dissatisfied': 3, 'Satisfied': 4, 'Very satisfied': 5}
# df['JobSat_score'] = df['JobSat'].map(jobsat_mapping)

# Define bins and labels for experience ranges
bins = [-1, 5, 10, 20, 100]
labels = ['0-5', '5-10', '10-20', '>20']
df['Experience_Range'] = pd.cut(df['YearsCodePro'], bins=bins, labels=labels)

# Calculate median JobSat score for each range (use 'JobSat' or your mapped numeric column)
median_satisfaction = df.groupby('Experience_Range', observed=False)['JobSat'].median() # Change to 'JobSat_score' if mapped
print(median_satisfaction)

# Visualize the relationship using a bar plot
plt.figure(figsize=(8, 5))
median_satisfaction.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Median Job Satisfaction by Years of Professional Experience')
plt.xlabel('Years of Experience Range')
plt.ylabel('Median Job Satisfaction')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

#### Step 5: Visualize Job Satisfaction


Use a count plot to show the distribution of `JobSat` values. This provides insights into the overall satisfaction levels of respondents.


```python
## Write your code here

# Visualize the distribution of JobSat values
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='JobSat', order=df['JobSat'].value_counts().index, palette='viridis')
plt.title('Distribution of Job Satisfaction')
plt.xlabel('Job Satisfaction')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

#### Step 6: Analyzing Remote Work Preferences by Job Role


Analyze trends in remote work based on job roles. Use the `RemoteWork` and `Employment` columns to explore preferences and examine if specific job roles prefer remote work more than others.

- Use a count plot to show remote work distribution.

- Cross-tabulate remote work preferences by employment type (e.g., full-time, part-time) and job roles.


```python
## Write your code here

# Use a count plot to show remote work distribution
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='RemoteWork', order=df['RemoteWork'].value_counts().index, palette='Set2')
plt.title('Distribution of Remote Work Preferences')
plt.xlabel('Remote Work Type')
plt.ylabel('Count')
plt.show()

# Cross-tabulate remote work preferences by employment type and job roles (assuming 'DevType' or 'JobRole' exists; adjust column name if needed)
role_col = 'DevType' if 'DevType' in df.columns else df.columns[0] # Fallback if column name varies
cross_tab = pd.crosstab(df['RemoteWork'], [df['Employment'], df.get('DevType', df['Employment'])])
print(cross_tab.head())
```

#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
## Write your code here

# Split language strings and explode them
languages = df['LanguageHaveWorkedWith'].dropna().str.split(';', expand=True).stack().reset_index(level=1, drop=True)
lang_df = df[['Country']].join(languages.rename('Language'))

# Clean country strings just in case of trailing spaces
lang_df['Country'] = lang_df['Country'].str.strip()

# Filter for the target country (Change 'United States' if your dataset uses a different format)
target_country = 'United States'
filtered_data = lang_df[lang_df['Country'] == target_country]['Language'].value_counts()

# Check if data exists before attempting to plot
if not filtered_data.empty:
    top_languages = filtered_data.head(10)
    
    plt.figure(figsize=(10, 5))
    top_languages.plot(kind='bar', color='coral', edgecolor='black')
    plt.title(f'Top 10 Programming Languages in {target_country}')
    plt.xlabel('Programming Language')
    plt.ylabel('Count')
    plt.xticks(rotation=45)
    plt.show()
else:
    print(f"No data found for country: '{target_country}'. Check your unique country values using df['Country'].unique().")

```

#### Step 8: Correlation Between Experience and Satisfaction


Examine how years of experience (`YearsCodePro`) correlate with job satisfaction (`JobSatPoints_1`). Use a scatter plot to visualize this relationship.


```python
## Write your code here

# Visualize relationship between YearsCodePro and JobSatPoints_1 using a scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='YearsCodePro', y='JobSatPoints_1', alpha=0.5, color='b')
plt.title('Correlation Between Professional Experience and Job Satisfaction Points')
plt.xlabel('Years of Professional Experience (YearsCodePro)')
plt.ylabel('Job Satisfaction Points')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

#### Step 9: Educational Background and Employment Type


Explore how educational background (`EdLevel`) relates to employment type (`Employment`). Use cross-tabulation and visualizations to understand if higher education correlates with specific employment types.


```python
## Write your code here

# Cross-tabulate educational background with employment type
ed_emp_crosstab = pd.crosstab(df['EdLevel'], df['Employment'], normalize='index') * 100
print(ed_emp_crosstab)

# Visualize using a stacked bar chart
plt.figure(figsize=(12, 6))
ed_emp_crosstab.plot(kind='bar', stacked=True, colormap='Spectral', edgecolor='black')
plt.title('Employment Type Breakdown by Educational Level (%)')
plt.xlabel('Educational Level')
plt.ylabel('Percentage')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
